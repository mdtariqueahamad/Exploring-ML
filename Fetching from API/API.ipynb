```python
import pandas as pd
import requests
```

```python
df = pd.DataFrame()
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
apikey = os.getenv("CRICAPI_KEY")
```

```python
for i in range (60):
    resp = requests.get(f"https://api.cricapi.com/v1/players?apikey={apikey}&offset={i}")
    df = pd.concat([df, pd.DataFrame(resp.json()['data']).iloc[:, 1:]], ignore_index=True)

df
```

```python
df.to_csv('cric_ply.csv')
```
